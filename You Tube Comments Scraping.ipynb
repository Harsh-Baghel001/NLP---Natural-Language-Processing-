```python
from youtube_comment_downloader import YoutubeCommentDownloader
import pandas as pd

# Video ID or full URL
video_id = 'GSnipy63VX4'  # e.g., 'dQw4w9WgXcQ'

downloader = YoutubeCommentDownloader()

# ✅ sort_by should be 0 for 'Top Comments'
comments = []
for comment in downloader.get_comments_from_url(f"https://www.youtube.com/watch?v={video_id}", sort_by=0):
    comments.append({
        'author': comment['author'],
        'text': comment['text'],
        'time': comment['time'],
        'likes': comment['votes']
    })

# Save to CSV
df = pd.DataFrame(comments)
df.to_csv('youtube_comments.csv', index=False)
print("Scraping completed. Total comments:", len(df))
```
